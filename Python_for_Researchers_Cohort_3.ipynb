## Dictionaries

```python
# Creating a dictionary
patient = {
    'name': "John Doe",
    'age': 56,
    'is_patient_ill': True,
    'weight': 75.5,
    'height': 1.75
}

print(patient)
type(patient)
```

```python
patient
```

```python
# Accessing a value from a dictionary
patient["name"]
```

```python
# Accessing a value from a dictionary
patient["names"]
```

```python
patient.get("age")
```
